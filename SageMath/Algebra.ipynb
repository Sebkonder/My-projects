
<strong>Defincja:</strong> Liczby $n,m\in \mathbb{N}$ nazywa się <em>zaprzyjaźnionymi</em> (ang. <em>amicable</em>), jeżeli suma dzielników właściwych (czyli różnych od $n$) liczby $n$ jest równa $m$, zaś suma dzielników właściwych (czyli różnych od $m$) liczby $m$ jest równa $n$.</p>
<p><em>Przykład:</em> Liczby 220, 284 są zaprzyjaźnione.</p>
<p><strong>Zadanie:</strong> Napisać funkcję, która sprawdza czy dwie podane liczby są zaprzyjaźnione.

```python
def czy_zaprzyjaznione(n, m):
    """
    Funkcja sprawdza czy podane liczby n, m są zaprzyjaźnione.
    """
    # wyznaczamy dzielniki WŁAŚCIWE liczby n
    dn = [ k for k in n.divisors() if k < n ]
    # jeżeli ich suma != m to kończymy
    if sum(dn) != m :
        return false
    # wyznaczamy dzielniki WŁAŚCIWE liczby m
    dm = [ k for k in m.divisors() if k < m ]
    # zwracamy wynik
    return sum(dm) == n
```

```python
# sprawdzamy na przykładach
print( czy_zaprzyjaznione(220, 284) )
print( czy_zaprzyjaznione(100485, 124155) )
```

```python
def Horner(f, a) :
    """
    Funkcja oblicza wartość f(a) za pomoca schematu Hornera.
    
    Argumenty:
        
        - f wielomian
        - a argument
    """
    # wynik będziemy akumulować w zmiennej "wynik"
    # na początku wstawiamy najwyższy współczynnik f
    wynik = f.leading_coefficient()
    # główna pętla
    for j in [ f.degree()-1, f.degree()-2..0 ] :
        # dotychczasowy wynik mnożymy przez a i dodajemy do niego j-ty współczynnik f
        wynik = f[j] + a*wynik
    # zwracamy wynik
    return wynik
```

<p><strong>Zadanie:</strong> Napisać funkcję, która oblicza NWD swoich argumentów za pomocą algorytmu Euklidesa.</p>

```python
def nwd_r( a, b ) :
    """
    Funkcja oblicza NWD(a,b) za pomocą algorytmu Euklidesa.
    """
    if b == 0 :
        return abs(a) if a in ZZ else a
    else :
        return nwd_r( b, a.mod(b) )
```

<p><strong>Zadanie:</strong> Zaimplementować metodę Lagrange'a rozwiązywania problemu Chińskiego o resztach.</p>

```python
def TCR( M, A ) :
    """
    Twierdzenie Chińskie o Resztach. Funkcja przyjmuje jako argumentu:
        
        - M - lista moduli
        
        - A - lista reszt
      
    W Wyniku funkcja zwraca element a t.ż. a przystaje do A[i] modulo M[i], dla wszystkich i
    """
    assert len(M) == len(A), "Listy M oraz A muszą mieć równe długości."
    assert all( gcd(M[i], M[j]) == 1 for i in range(len(M)) for j in range(len(M)) if i != j ), "Moduli muszą być względnie pierwsze."
    
    # zapisujemy liczbę reszt
    k = len(A)
    # wyznaczamy współczynniki Bezout
    alpha = zero_matrix(k)
    for i in [0..k-1] :
        for j in [i+1..k-1] :
            _, a, b = xgcd(M[i], M[j])
            alpha[i,j] = a
            alpha[j,i] = b
            
    # konstruujemy elementy odpowiednie iloczyny
    B = [ prod(alpha[j,i]*M[j] for j in [0..k-1] if j != i) for i in [0..k-1] ]
    
    # konstruujemy element wynikowy
    a = sum( A[i]*B[i] for i in [0..k-1] ) 
    # zwracamy wynik
    return a.mod( prod(M) )
```

```python
# sprawdzenie (przykład z LCM)
TCR([5,7,9], [1,6,5])
```

<p><strong>Zadanie:</strong> Zaimplementować algorytm pomocniczy do metody Newtona rozwiązywania problemu Chińskiego o resztach</p>

```python
def TCR2( m1, m2, a1, a2 ) :
    """
    Funkcja przyjmuje jako argumenty:
        - moduli m1, m2
        - reszty a1, a2
    W wyniku zwraca a t.ż. a przystaje do a1 modulo m1 oraz do a2 modulo m2
    """

    assert gcd(m1, m2) == 1, "Mm1, m2 muszą być względnie pierwsze"
    
    _, alpha, beta = xgcd(m1, m2)
    r1 = a1.mod(m1)
    b = (a2 - r1)*alpha
    r2 = b.mod(m2)
    return r1 + r2*m1
```

<p><strong>Zadanie:</strong> Zaimplementować metodę Newtona rozwiązywania problemy Chińskiego o resztach</p>

```python
def TCR( M, A ) :
    """
    Twierdzenie Chińskie o Resztach. Funkcja przyjmuje jako argumentu:
        
        - M - lista moduli
        
        - A - lista reszt
      
    W wyniku funkcja zwraca element a t.ż. a przystaje do A[i] modulo M[i], dla wszystkich i
    """
    assert len(M) == len(A), "Listy M oraz A muszą mieć równe długości."
    assert all( gcd(M[i], M[j]) == 1 for i in range(len(M)) for j in range(len(M)) if i != j ), "Moduli muszą być względnie pierwsze."

    # zapisujemy liczbę reszt
    k = len(A)
    # inicjalizacja
    m = M[0]
    a = A[0]
    
    # główna pętla
    for i in [1..k-1] :
        a = TCR2(m, M[i], a, A[i])
        m *= M[i]
        
    # zwracamy wynik
    return a
```

```python
# sprawdzenie (przykład z LCM)
TCR([5,7,9], [1,6,5])
```

```python
def potega( a, n ) :
    """
    Funkcja oblicza n-tą potęgę argumentu a, za pomocą podnoszenia do kwadratu.
    """
    assert n in NN, "Wykładnik musi być liczbą naturalną!"
    
    r = 1 # tu będziemy przechowywać wynik
    b = a # tu będą kolejne kwadraty
    while n != 0 :
        if is_odd(n) :
            r *= b
        b *= b
        n //= 2
    return r
```

```python
def fi( n ) :
    """
    Funkcja wyznacza `\\varphi(n)`
    """
    
    F = factor(n)
    return prod( p^k - p^(k-1) for (p,k) in F )
```

<strong>Zadanie:</strong> Napisać funkcję, która oblicza $a^n\bmod{m}$

```python
def potega_mod( a, n, m ) :
    """
    Funkcja oblicza n-tą potęgę argumentu liczba a modulo m.
    """
    assert n in NN, "Wykładnik musi być liczbą naturalną!"
    
    # redukujemy wykładnik modulo fi(m)
    n = n.mod( fi(m) )
    # końcówkę obliczamy poprzez podnoszenie do kwadratu
    Zm = Zmod(m)
    r = potega( Zm(a), n )
    return ZZ(r)
```

<p><strong>Zadanie:</strong> Zaimplementować algorytm modularnego obliczania wyznacznika macierzy całkowitej.</p>

```python
def det_mod( M ) :
    """
    Funkcja oblicza wyznacznik macierzy o współczynnikach całkowitych.
    """
    assert M.is_square(), "Macierz musi być kwadratowa!"
    
    # wyznaczamy ograniczenie górne
    B = max(map(abs, M.list()))
    # oszacowanie Hadamarda
    n = M.nrows()
    H = 2*B^n*ceil(sqrt(n^n))
    
    P = []
    D = []
    m = 1
    # pętla po liczbach pierwszych
    p = 2
    while m <= H :
        p = next_prime(p)
        P.append(p)
        D.append(ZZ( M.change_ring(GF(p)).det() ))
        m *= p
    # twierdzenie Chińskie o resztach
    d = TCR( P, list(map(ZZ,D)))
    if d > m/p :
        d -= m
    # zwracamy wynik
    return d
```
