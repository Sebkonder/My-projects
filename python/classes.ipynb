## `Osoba` – walidacja numeru PESEL
Modeluje osobę i sprawdza poprawność numeru PESEL (suma kontrolna) oraz wylicza z niego datę urodzenia. Pola są chronione właściwościami z walidacją.

```python
class Osoba1:
  '''
  Klasa opisuje osobę i waliduje numer PESEL.

  Args:
    imie (str): imię
    nazwisko (str): nazwisko
    pesel (str): 11-cyfrowy numer PESEL

  Attrib:
    imie, nazwisko, pesel: walidowane właściwości
    data (str): data urodzenia wyliczona z numeru PESEL
  '''
  def __init__(self, imie, nazwisko, pesel):
    self.imie = imie
    self.nazwisko = nazwisko
    self.pesel = pesel
    if pesel[2] in {'0', '1'}:
      self.data = '19' + pesel[:2] + '/' + pesel[2:4] + '/' + pesel[4:6]
    elif pesel[2] in {'2', '3'}:
      self.data = '20' + pesel[:2] + '/' + str(int(pesel[2:4]) - 20) + '/' + pesel[4:6]

  def __str__(self):
    ret = f'Imie: {self.__imie}'
    ret += f'\nNazwisko: {self.__nazwisko}'
    ret += f'\nPesel: {self.__pesel}'
    ret += f'\nData urodzenia: {self.data}\n'
    return ret

  @property
  def imie(self):
    return self.__imie

  @imie.setter
  def imie(self, nowe_imie):
    if not isinstance(nowe_imie, str):
      raise TypeError('Podaj napis!')
    if not nowe_imie.isalpha():
      raise ValueError('Imie zawiera niedozwolone znaki')
    self.__imie = nowe_imie

  @property
  def nazwisko(self):
    return self.__nazwisko

  @nazwisko.setter
  def nazwisko(self, nowe_nazwisko):
    if not isinstance(nowe_nazwisko, str):
      raise TypeError('Podaj napis!')
    if not nowe_nazwisko.isalpha():
      raise ValueError('Nazwisko zawiera niedozwolone znaki')
    self.__nazwisko = nowe_nazwisko

  @property
  def pesel(self):
    return self.__pesel

  @pesel.setter
  def pesel(self, nowy_pesel):
    if not isinstance(nowy_pesel, str):
      raise TypeError('Podaj napis!')
    if not nowy_pesel.isdigit():
      raise ValueError('Pesel zawiera litery')
    if len(nowy_pesel) != 11:
      raise ValueError('Pesel musi miec 11 cyfr')

    wagi = [1, 3, 7, 9, 1, 3, 7, 9, 1, 3]
    suma = 0
    for i in range(len(wagi)):
      suma += (int(nowy_pesel[i]) * wagi[i]) % 10
    suma = 10 - suma % 10

    if suma == int(nowy_pesel[-1]):
      self.__pesel = nowy_pesel
    else:
      raise ValueError('Pesel niepoprawny')

  def __lt__(self, other):
    return self.nazwisko < other.nazwisko
```

```python
o1 = Osoba1('Mike', 'Kogh', '01250905051')
o2 = Osoba1('John', 'Koghy', '82060821869')
print(o1)
print(o2)
print('o1 < o2 (po nazwisku):', o1 < o2)
```

## `Vector` – wektor 3D z przeciążonymi operatorami\nDodawanie, odejmowanie, mnożenie przez skalar, iloczyn skalarny i wektorowy, długość oraz kąt między wektorami.

```python
import math

class Vector:
  '''
  Wektor 3-wymiarowy z przeciążonymi operatorami.

  Obsługuje: dodawanie, odejmowanie, mnożenie przez skalar,
  iloczyn skalarny, porównanie, indeksowanie, długość,
  iloczyn wektorowy oraz kąt między wektorami.
  '''
  def __init__(self, x1, x2, x3):
    self.x1 = x1
    self.x2 = x2
    self.x3 = x3

  def __str__(self):
    return f'[{self.x1}, {self.x2}, {self.x3}]'

  def __repr__(self):
    return self.__str__()

  @property
  def x1(self):
    return self.__x1

  @x1.setter
  def x1(self, value):
    if isinstance(value, (int, float)):
      self.__x1 = value
    else:
      raise ValueError('Koordynat musi byc liczba')

  @property
  def x2(self):
    return self.__x2

  @x2.setter
  def x2(self, value):
    if isinstance(value, (int, float)):
      self.__x2 = value
    else:
      raise ValueError('Koordynat musi byc liczba')

  @property
  def x3(self):
    return self.__x3

  @x3.setter
  def x3(self, value):
    if isinstance(value, (int, float)):
      self.__x3 = value
    else:
      raise ValueError('Koordynat musi byc liczba')

  def __add__(self, other):
    return Vector(self.x1 + other.x1, self.x2 + other.x2, self.x3 + other.x3)

  def __sub__(self, other):
    return Vector(self.x1 - other.x1, self.x2 - other.x2, self.x3 - other.x3)

  def __rmul__(self, a):
    return Vector(a * self.x1, a * self.x2, a * self.x3)

  def __mul__(self, other):
    if isinstance(other, Vector):
      return self.x1 * other.x1 + self.x2 * other.x2 + self.x3 * other.x3
    return Vector(self.x1 * other, self.x2 * other, self.x3 * other)

  def __eq__(self, other):
    if not isinstance(other, Vector):
      raise TypeError('Musisz podac obiekt typu Vector')
    return self.x1 == other.x1 and self.x2 == other.x2 and self.x3 == other.x3

  def __getitem__(self, i):
    return [self.x1, self.x2, self.x3][i]

  def dlugosc(self):
    return (self.x1 ** 2 + self.x2 ** 2 + self.x3 ** 2) ** 0.5

  def iloczyn_wekt(self, other):
    if not isinstance(other, Vector):
      raise TypeError('Musisz podac obiekt typu Vector')
    return Vector(self.x2 * other.x3 - self.x3 * other.x2,
                  self.x3 * other.x1 - self.x1 * other.x3,
                  self.x1 * other.x2 - self.x2 * other.x1)

  def kat(self, other):
    if not isinstance(other, Vector):
      raise TypeError('Musisz podac obiekt typu Vector')
    return math.acos((self * other) / (self.dlugosc() * other.dlugosc()))
```

```python
v1 = Vector(1, 2, 3)
v2 = Vector(4, 5, 6)

print('v1 + v2      =', v1 + v2)
print('v1 . v2      =', v1 * v2)          # iloczyn skalarny
print('v1 x v2      =', v1.iloczyn_wekt(v2))  # iloczyn wektorowy
print('|v1|         =', v1.dlugosc())
print('kat (rad)    =', v1.kat(v2))
```

## `Fib` – iterator Fibonacciego

```python
class Fib:
  '''Iterator zwracajacy liczby Fibonacciego do zadanej granicy.'''
  def __init__(self, end=10):
    self.x = 0
    self.y = 1
    self.end = end

  def __iter__(self):
    return self

  def __next__(self):
    if self.x > self.end:
      raise StopIteration
    temp = self.x
    self.x, self.y = self.y, self.x + self.y
    return temp
```

```python
for n in Fib(10):
  print(n, end=' ')
```

## `factorial` – funkcja z testami `doctest`

```python
def factorial(n):
  """
  Zwraca silnie liczby calkowitej nieujemnej.

  Args:
    n (int): liczba calkowita >= 0
  Returns:
    int: silnia n

  >>> factorial(4)
  24
  >>> factorial(0)
  1
  >>> factorial(-4)
  Traceback (most recent call last):
    ...
  ValueError: Liczba musi byc dodatnia
  >>> factorial(4.5)
  Traceback (most recent call last):
    ...
  ValueError: Liczba musi byc dodatnia
  """
  if n < 0:
    raise ValueError('Liczba musi byc dodatnia')
  if n != int(n):
    raise ValueError('Liczba musi byc dodatnia')
  wynik = 1
  while n > 0:
    wynik = wynik * n
    n -= 1
  return wynik
```

```python
import doctest
doctest.testmod(verbose=False)
```
