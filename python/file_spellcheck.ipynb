# Przetwarzanie tekstu\n\n## `Plik` – poprawa pisowni w pliku tekstowym\nKlasa wczytuje plik tekstowy i interaktywnie proponuje zamianę częstego polskiego wzorca pisowni (**uw → ów**), z listą wyjątków typu *skuwka*, *wsuwka*, *zasuwka*.

````python
class Plik:
  '''
  Klasa Plik obsługuje pliki tekstowe z możliwością poprawy pisowni.

  Args:
    path (str): ścieżka do pliku

  Methods:
    wczytaj: otwiera plik pod podaną ścieżką
    zapisz: zamyka wcześniej otwarty plik
    sprawdz: interaktywnie poprawia wzorzec "uw" -> "ów" (z listą wyjątków)
    wypisz_dane_o_pliku: wypisuje ścieżkę i liczbę słów
    wypisz_zawartosc: wypisuje zawartość pliku
  '''
  def __init__(self, path):
    self.path = path
    self.f = None

  def wczytaj(self):
    self.f = open(self.path, 'r+')

  def zapisz(self):
    try:
      self.f.close()
    except Exception:
      print('Plik nie jest otwarty!')

  def sprawdz(self):
    wyjatki = ['zasuwka', 'skuwka', 'wsuwka']
    if self.f is None:
      raise Exception('Wczytaj plik!')

    self.f.seek(0)
    text = self.f.read()

    for word in set(text.split()):
      if 'uw' in word and word not in wyjatki:
        opt = input(f'Czy chcesz poprawic uw w "{word}" na ow? [T/N] ')
        if opt == 'T':
          text = text.replace(word, word.replace('uw', 'ow'))

    if input('Czy chcesz zapisac zmiany? [T/N] ') == 'T':
      self.f.seek(0)
      self.f.write(text)
      self.f.truncate()

  def wypisz_dane_o_pliku(self):
    if self.f is None:
      raise Exception('Wczytaj plik!')
    self.f.seek(0)
    print(f'Sciezka do pliku: {self.path}')
    print(f'Liczba slow: {len(self.f.read().split())}')

  def wypisz_zawartosc(self):
    if self.f is None:
      raise Exception('Wczytaj plik!')
    self.f.seek(0)
    print(self.f.read())
````

### Przykład użycia\n```python\nf = Plik('text.txt')\nf.wczytaj()\nf.wypisz_dane_o_pliku()\nf.sprawdz()        # zadaje pytania w konsoli\nf.zapisz()\n```\n*Metoda `sprawdz` korzysta z `input()`, więc działa interaktywnie (np. w konsoli lub notebooku), a nie w trybie wsadowym.*
