## Lista wiązana (`LinkedList`)
Własna jednokierunkowa lista z operacjami `append`, `pop`, `isEmpty` i czytelną reprezentacją tekstową.

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

class LinkedList:
  '''Jednokierunkowa lista wiazana.'''
  def __init__(self):
    self.head = None
    self.tail = None

  def append(self, data):
    new_node = Node(data)
    if self.head is None:
      self.head = new_node
      self.tail = new_node
    else:
      self.tail.next = new_node
      self.tail = new_node

  def isEmpty(self):
    return self.head is None

  def pop(self):
    if self.isEmpty():
      return None
    temp = self.head
    self.head = self.head.next
    return temp.data

  def __str__(self):
    current = self.head
    result = ""
    while current is not None:
      result += str(current.data) + " -> "
      current = current.next
    return result + "None"
```

```python
l = LinkedList()
l.append(10)
l.append(11)
print(l)
l.pop()
l.append(20)
print(l)
```

## Graf (`Graph`)
Graf skierowany na liście sąsiedztwa zbudowanej na `defaultdict(set)`.

```python
from collections import defaultdict

class Graph:
  '''Prosty graf skierowany oparty na liscie sasiedztwa.'''
  def __init__(self, connections):
    # defaultdict(set) tworzy pusty zbior dla nowego wierzcholka,
    # dzieki czemu .add() dziala bez wczesniejszej inicjalizacji klucza
    self._connections = defaultdict(set)
    self.add_connections(connections)

  def add_connections(self, connections):
    for vertex1, vertex2 in connections:
      self._connections[vertex1].add(vertex2)

  def __str__(self):
    return '{}({})'.format(self.__class__.__name__, dict(self._connections))
```

```python
g = Graph([("A", "B"), ("B", "C"), ("C", "D"), ("B", "A")])
print(g)
```
